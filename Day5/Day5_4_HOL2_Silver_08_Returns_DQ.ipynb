## Step 1 — Setup

## Step 3 — Investigate: Does `RefundAmount` Follow `ReturnStatus`?
Sample rows suggest `RefundAmount = 0.0` for `Pending`/`Rejected` and a
real amount for `Approved`/`Refunded`. Check the full table, in both
directions, before assuming the pattern is clean.

```python
bronze_df.groupBy("ReturnStatus") \
    .agg(
        count(when(col("RefundAmount") == 0.0, 1)).alias("amount_zero"),
        count(when(col("RefundAmount") > 0.0, 1)).alias("amount_positive"),
        count(when(col("RefundAmount").isNull(), 1)).alias("amount_null"),
        count("*").alias("total")
    ).display()
```

### Finding: RefundAmount Follows ReturnStatus Exactly

| ReturnStatus | amount_zero | amount_positive | amount_null | total |
|---|---|---|---|---|
| Pending | 4,174 | 0 | 0 | 4,174 |
| Approved | 0 | 12,439 | 0 | 12,439 |
| Refunded | 0 | 8,227 | 0 | 8,227 |
| Rejected | 4,160 | 0 | 0 | 4,160 |

Every `Pending`/`Rejected` return has `RefundAmount = 0.0` — no refund has
been issued yet, or none ever will be. Every `Approved`/`Refunded` return
has a real, positive `RefundAmount`. **0 violations, 0 nulls, across all
28,964 rows.**

### What This Looks Like for Real Customers

**Pending** — *Priya ordered a blender. It arrived damaged, so she
clicked "Return this item" yesterday.* GlobalMart hasn't reviewed her
request yet, so no money has gone back to her.
`RefundAmount = 0.0` — nothing has happened yet, only a request was made.

**Approved** — *Rahul returned a jacket that didn't fit. GlobalMart
reviewed and approved the return.* The refund amount is now set — even
if the money hasn't landed in his account yet.
`RefundAmount = 2500.0` — the refund has officially been sanctioned.

**Refunded** — *Amit returned headphones a week ago. GlobalMart approved
it, and the money has now actually been credited back to him.*
`RefundAmount = 1800.0` — one step further than Approved: money moved.

**Rejected** — *Sneha tried to return used shoes after 60 days, past the
return window. GlobalMart said no.* She will never get money back.
`RefundAmount = 0.0` — not "not yet," but "never will be."

This is a deliberate business rule, not missing or broken data — same
pattern as `material` nulls in `products` and gift card/coupon amounts in
`payments`. Summing `RefundAmount` for "total refunded to customers" is
already correct as-is, with no filtering needed.

**Decision: no fix, no quarantine for this pattern.**

## Step 4 — DQ Scan

| Rule | What we check | Why |
|---|---|---|
| `NULL_RETURN_ID` | `ReturnID` is null | Primary key |
| `NULL_ORDER_ID` | `OrderId` is null | FK to `silver.orders` |
| `NULL_PRODUCT_ID` | `ProductID` is null | FK to `silver.products` |
| `NULL_RETURN_DATE` | `ReturnDate` is null | Needed for `dim_date` join in Gold |
| `NEGATIVE_REFUND_AMOUNT` | `RefundAmount` < 0 | Can't refund a negative amount |

```python
dq_scan_df = bronze_df \
    .withColumn("_dq_issue",
        when(col("ReturnID").isNull(),                          lit("NULL_RETURN_ID"))
        .when(col("OrderId").isNull(),                          lit("NULL_ORDER_ID"))
        .when(col("ProductID").isNull(),                        lit("NULL_PRODUCT_ID"))
        .when(col("ReturnDate").isNull(),                       lit("NULL_RETURN_DATE"))
        .when(col("RefundAmount") < 0,                          lit("NEGATIVE_REFUND_AMOUNT"))
        .otherwise(lit(None))
    )

print("=== DQ Issues Found ===")
dq_scan_df.groupBy("_dq_issue").count().orderBy("count", ascending=False).display()
```

```python
# ReturnID uniqueness
dupe_count = bronze_df.groupBy("ReturnID").count().filter("count > 1").count()
print(f"Duplicate ReturnID count: {dupe_count}")

# Confirm actual distinct ReturnStatus / ReturnMethod values (don't assume)
bronze_df.groupBy("ReturnStatus").count().display()
bronze_df.groupBy("ReturnMethod").count().display()
```

**Finding: 0 DQ issues, 0 duplicate `ReturnID`.** 4 valid `ReturnStatus`
values (matching Step 3 exactly) and 3 `ReturnMethod` values (`Courier`,
`Pickup`, `Store_Drop`). `Store_Drop` has the same underscore-as-space
pattern seen in addresses' `State`/`AddressType` — cleaned in transform.

## Step 5 — Referential Integrity Check
`silver.orders` and `silver.products` already exist — check for orphans
before writing.

```python
silver_orders_df   = spark.table("harsh_kumar01_npmentorskool_onmicrosoft_com.silver.orders")
silver_products_df = spark.table("harsh_kumar01_npmentorskool_onmicrosoft_com.silver.products")

orphan_orders = bronze_df.join(
    silver_orders_df.select("order_id"), bronze_df.OrderId == col("order_id"), "left_anti"
)
orphan_products = bronze_df.join(
    silver_products_df.select("product_id"), bronze_df.ProductID == col("product_id"), "left_anti"
)

print(f"returns with no matching order_id  : {orphan_orders.count():,} / {bronze_df.count():,}")
print(f"returns with no matching product_id: {orphan_products.count():,} / {bronze_df.count():,}")
```

**Finding: 0 orphans on both checks.** Unlike `order_items`, this table
never ran into the orders-quarantine cascade, since the orders notebook
was already fixed (flag, not quarantine) by the time we got here.

## Step 6 — Transform & Write
No DQ issues found — no quarantine table needed. Two cleanups before
renaming: fix `Store_Drop` → `Store Drop`, and standardize the
inconsistent `OrderId`/`Return_reason` casing to snake_case.

```python
silver_df = bronze_df \
    .withColumn("ReturnMethod", regexp_replace(col("ReturnMethod"), r"_", " ")) \
    .withColumnRenamed("ReturnID", "return_id") \
    .withColumnRenamed("OrderId", "order_id") \
    .withColumnRenamed("ProductID", "product_id") \
    .withColumnRenamed("Return_reason", "return_reason") \
    .withColumnRenamed("ReturnDate", "return_date") \
    .withColumnRenamed("ReturnStatus", "return_status") \
    .withColumnRenamed("RefundAmount", "refund_amount") \
    .withColumnRenamed("ReturnMethod", "return_method") \
    .withColumn("_silver_updated_at", current_timestamp()) \
    .select(
        "return_id", "order_id", "product_id", "return_reason", "return_date",
        "return_status", "refund_amount", "return_method",
        "_source_file", "_silver_updated_at"
    )

silver_df.display()
```

```python
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.silver")

silver_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(SILVER_TABLE)

print(f"Written to {SILVER_TABLE}: {spark.table(SILVER_TABLE).count():,} rows")
```

## Step 7 — Verify

```python
df = spark.table(SILVER_TABLE)
print(f"silver.returns rows : {df.count():,}")
df.printSchema()

df.groupBy("return_status").count().display()
df.groupBy("return_method").count().display()
df.select(avg("refund_amount").alias("avg_refund_amount")).display()
```
