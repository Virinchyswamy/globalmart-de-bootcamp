## Step 2 — Read Raw Data from Bronze
Look at the actual schema and sample rows before assuming anything.

```python
bronze_df = spark.table(BRONZE_TABLE)
print(f"Total records in Bronze: {bronze_df.count():,}")
bronze_df.printSchema()
```

```python
bronze_df.display()
```

## Step 3 — Investigate: PinCode Leading-Zero Loss

`PinCode` was inferred as a numeric type by Autoloader. Indian PIN codes
are always 6 digits — if any originally started with `0`, casting to a
number has already destroyed that digit before we ever see it. Check how
many rows are affected, and whether it's really just one lost digit.

```python
pincode_check_df = bronze_df.withColumn(
    "_pincode_len", length(col("PinCode").cast("long").cast("string"))
)

short_count = pincode_check_df.filter(col("_pincode_len") < 6).count()
print(f"Rows with PinCode shorter than 6 digits: {short_count:,} / {bronze_df.count():,}")

pincode_check_df.filter(col("_pincode_len") < 6) \
    .groupBy("_pincode_len").count().orderBy("_pincode_len").display()
```

**Finding:** 2,861 rows affected, distributed `1: 1, 2: 3, 3: 33, 4: 263,
5: 2,561` — a clean ~10x pattern per digit, consistent with leading zeros
being stripped to different degrees depending on how many a PinCode
originally had.

## Step 4 — Confirm Root Cause: Recover PinCode from AddressLine1

Checking the raw source CSV directly (bypassing Spark's numeric
inference) shows the leading-zero loss already exists in the `PinCode`
column **before** ingestion — a source-side issue, not an Autoloader bug.

However, `AddressLine1` embeds the same PinCode as plain text at the end
of the string (e.g. `Guntur_082130`), and text concatenation never strips
leading zeros. That gives us an independent, trustworthy second copy of
the true value — no guessing required.

```python
extracted_df = bronze_df.withColumn(
    "_extracted_pincode", regexp_extract(col("AddressLine1"), r"(\d{5,6})\s*$", 1)
).withColumn(
    "_pincode_len", length(col("PinCode").cast("long").cast("string"))
)

short_df = extracted_df.filter(col("_pincode_len") < 6)

total_short        = short_df.count()
extraction_failed  = short_df.filter((col("_extracted_pincode") == "") | col("_extracted_pincode").isNull()).count()
mismatch_count     = short_df.filter(~col("_extracted_pincode").endswith(col("PinCode").cast("long").cast("string"))).count()

print(f"Total short-PinCode rows  : {total_short:,}")
print(f"Extraction failed (empty): {extraction_failed:,}")
print(f"Extracted/PinCode mismatch: {mismatch_count:,}")

short_df.select("AddressID", "AddressLine1", "PinCode", "_extracted_pincode").show(10, truncate=False)
```

**Finding:** 0 extraction failures, 0 mismatches across all 2,861 rows —
full confidence the value extracted from `AddressLine1` is the true
original PinCode.

**Decision:** use the extracted value to recover PinCode where short,
leave the other 27,139 rows untouched (just normalize away the `.0`).

```python
remediated_df = bronze_df.withColumn(
    "_extracted_pincode", regexp_extract(col("AddressLine1"), r"(\d{5,6})\s*$", 1)
).withColumn(
    "_pincode_len", length(col("PinCode").cast("long").cast("string"))
).withColumn(
    "PinCode",
    when(col("_pincode_len") < 6, col("_extracted_pincode"))
    .otherwise(col("PinCode").cast("long").cast("string"))
).drop("_extracted_pincode", "_pincode_len")

# Verify: every PinCode should now be exactly 6 characters
remediated_df.select(length(col("PinCode")).alias("len")).groupBy("len").count().display()
```

## Step 5 — Is a PinCode Starting with `0` Valid?

**Geographically, no** — real Indian PIN codes use the first digit for a
postal zone (1–9; there's no zone `0`), so values like `082130` could
never be a real, postable PIN code.

**But that's a different question from source fidelity.** We've cross-
validated all 2,861 short PinCodes against the independently-generated
text inside `AddressLine1` — 0 extraction failures, 0 mismatches. This is
exactly what the source system recorded, not a guess or casting artifact.

We also never validate any other PinCode against true city/state geography
in this dataset (e.g. nobody checks if `143611` is genuinely Mirzapur's
real PIN). Quarantining only the "starts with 0" rows on geographic-
validity grounds would be an inconsistent rule — enforcing real-world
correctness for one digit position while ignoring it everywhere else.

**Decision:** Keep all recovered PinCodes, including those starting with
`0`. Silver's job here is *source fidelity*, not India Post zone
validation — that would be a separate, explicit rule with its own
reference table, out of scope for this notebook.

## Step 6 — Investigate & Fix: Duplicate City/PinCode in AddressLine1

`AddressLine1` embeds the street address PLUS a duplicate of City/PinCode
at the end (e.g. `"61/96,_Chauhan_Road\nHaldia_060162"`), joined with
inconsistent separators (`\n`, `_`, `-`, comma). Since `City` and
`PinCode` already exist as clean, separate columns, this trailing portion
is pure duplication.

Roughly 2 of every 3 rows (19,088/30,000) use a line-break separator; the
rest use a comma — same duplication pattern, different punctuation.

```python
multiline_count = bronze_df.filter(col("AddressLine1").contains("\n")).count()
print(f"Rows with embedded newline in AddressLine1: {multiline_count:,} / {bronze_df.count():,}")

bronze_df.filter(col("AddressLine1").contains("\n")) \
    .select("AddressID", "AddressLine1", "City", "PinCode").show(10, truncate=False)

bronze_df.filter(~col("AddressLine1").contains("\n")) \
    .select("AddressID", "AddressLine1", "City", "PinCode").show(10, truncate=False)
```

### Fix

The duplicate chunk always starts right after the **last** comma or line
break in the string — whichever comes later. Everything before that point
is the real street address; everything after is the repeat we don't need.

`[,\n][^,\n]*$` matches starting at a comma-or-newline only if everything
*after* it contains no other comma or newline — that constraint forces
the match onto the last separator automatically, whether it's a comma, a
single line break, or the second of two line breaks.

After removing it, we also clean up leftover formatting noise: `_`
standing in for spaces, and any remaining line breaks from multi-line
street descriptions (e.g. house number on one line, street name on the
next).

```python
cleaned_df = remediated_df \
    .withColumn("AddressLine1", regexp_replace(col("AddressLine1"), r"[,\n][^,\n]*$", "")) \
    .withColumn("AddressLine1", regexp_replace(col("AddressLine1"), r"\n", ", ")) \
    .withColumn("AddressLine1", regexp_replace(col("AddressLine1"), r"_", " ")) \
    .withColumn("AddressLine1", trim(col("AddressLine1")))

cleaned_df.select("AddressID", "AddressLine1", "City", "PinCode").show(15, truncate=False)
```

## Step 7 — Clean `State` and `AddressType`
Same underscore-as-space pattern as `AddressLine1` — low-risk cosmetic fix.

```python
cleaned_df = cleaned_df \
    .withColumn("State", regexp_replace(col("State"), r"_", " ")) \
    .withColumn("AddressType", regexp_replace(col("AddressType"), r"_", " "))

cleaned_df.select("AddressID", "City", "State", "AddressType").distinct().show(10, truncate=False)
```

## Step 8 — DQ Scan

| Check | Result |
|---|---|
| Distinct `AddressType` values | `Shipping`, `Billing`, `Billing, Shipping` — 10,000 each, no unexpected values |
| `AddressID` duplicates | 0 |
| Nulls on `AddressID`/`CustomerID`/`City`/`State` | 0 |

```python
cleaned_df.groupBy("AddressType").count().orderBy("count", ascending=False).display()

dupe_count = cleaned_df.groupBy("AddressID").count().filter("count > 1").count()
print(f"Duplicate AddressID count: {dupe_count}")

cleaned_df.select(
    count(when(col("AddressID").isNull(), 1)).alias("null_address_id"),
    count(when(col("CustomerID").isNull(), 1)).alias("null_customer_id"),
    count(when(col("City").isNull(), 1)).alias("null_city"),
    count(when(col("State").isNull(), 1)).alias("null_state")
).display()
```

**Finding: 0 DQ issues.** Like `products`, this dataset needs no
quarantine table — `cleaned_df` is already our clean dataset.

## Step 9 — Transform & Write

```python
silver_df = cleaned_df \
    .withColumnRenamed("AddressID", "address_id") \
    .withColumnRenamed("CustomerID", "customer_id") \
    .withColumnRenamed("AddressLine1", "address_line1") \
    .withColumnRenamed("City", "city") \
    .withColumnRenamed("State", "state") \
    .withColumnRenamed("PinCode", "pincode") \
    .withColumnRenamed("AddressType", "address_type") \
    .withColumn("_silver_updated_at", current_timestamp()) \
    .select(
        "address_id", "customer_id", "address_line1", "city", "state",
        "pincode", "address_type", "_source_file", "_silver_updated_at"
    )

silver_df.display()
```

```python
# spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.silver")

silver_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(SILVER_TABLE)

print(f"Written to {SILVER_TABLE}: {spark.table(SILVER_TABLE).count():,} rows")
```

## Step 10 — Verify

```python
df = spark.table(SILVER_TABLE)
print(f"silver.address rows : {df.count():,}")
df.printSchema()

df.select(length(col("pincode")).alias("len")).groupBy("len").count().display()
df.groupBy("address_type").count().display()
df.select("address_id", "address_line1", "city", "state", "pincode").show(10, truncate=False)
```
