## Step 1 — Setup

````python
from pyspark.sql.functions import *
from pyspark.sql.types import DateType
from delta.tables import DeltaTable

CATALOG      = "harsh_kumar01_npmentorskool_onmicrosoft_com"
BRONZE_TABLE = "harsh_kumar01_npmentorskool_onmicrosoft_com.bronze.orders"
SILVER_TABLE = "harsh_kumar01_npmentorskool_onmicrosoft_com.silver.orders"

# Actual channel values confirmed in the data (Step 7)
VALID_CHANNELS = ["Online", "Retail PoS"]

print(f"Reading from : {BRONZE_TABLE}")
print(f"Writing to   : {SILVER_TABLE}")
````

## Step 2 — Read Raw Data from Bronze
Orders are ingested via **Lakeflow Connect CDC** from Supabase Postgres.
Bronze holds the latest snapshot of each order.

> Lakeflow Connect lowercases all column names from Postgres —
> `OrderID` → `orderid`, `CustomerID` → `customerid`, etc.

````python
bronze_df = spark.table(BRONZE_TABLE)
print(f"Total records in Bronze: {bronze_df.count():,}")
print(f"Columns: {bronze_df.columns}")
bronze_df.display()
````

## Step 3 — First DQ Scan
Tag every row with its first failing rule.

| Rule | What we check |
|---|---|
| `NULL_ORDER_ID` | orderid is null |
| `NULL_CUSTOMER_ID` | customerid is null |
| `NULL_ORDER_DATE` | orderdate is null |
| `SHIP_BEFORE_ORDER` | shippingdate < orderdate |
| `DELIVERY_BEFORE_SHIP` | actualdeliverydate < shippingdate |
| `FUTURE_ORDER_DATE` | orderdate > current timestamp |
| `INVALID_ORDER_CHANNEL` | channel not in `[Online, Retail PoS]` |

This first pass compares **raw timestamps** — we'll see why that matters.

````python
dq_scan_df = bronze_df \
    .withColumn("_dq_issue",
        when(col("orderid").isNull(),                                       lit("NULL_ORDER_ID"))
        .when(col("customerid").isNull(),                                   lit("NULL_CUSTOMER_ID"))
        .when(col("orderdate").isNull(),                                    lit("NULL_ORDER_DATE"))
        .when(
            col("shippingdate").isNotNull() & (col("shippingdate") < col("orderdate")),
            lit("SHIP_BEFORE_ORDER")
        )
        .when(
            col("actualdeliverydate").isNotNull() & col("shippingdate").isNotNull() &
            (col("actualdeliverydate") < col("shippingdate")),
            lit("DELIVERY_BEFORE_SHIP")
        )
        .when(col("orderdate") > current_timestamp(),                       lit("FUTURE_ORDER_DATE"))
        .when(~col("orderchannel").isin(VALID_CHANNELS),                    lit("INVALID_ORDER_CHANNEL"))
        .otherwise(lit(None))
    )

print("=== DQ Issues Found (timestamp-level) ===")
dq_scan_df.groupBy("_dq_issue").count().orderBy("count", ascending=False).display()
````

## Step 4 — Investigate: DELIVERY_BEFORE_SHIP
725 orders flagged. Look at the actual records before deciding anything.

````python
dq_scan_df.filter(col("_dq_issue") == "DELIVERY_BEFORE_SHIP") \
    .select(
        "orderid", "customerid", "shippingdate", "actualdeliverydate",
        datediff(col("shippingdate").cast(DateType()), col("actualdeliverydate").cast(DateType())).alias("days_diff")
    ) \
    .display()
````

### Why timestamp comparison is misleading

Most of the 725 flagged rows show `days_diff` of **0 or 1** — not the
multi-day gaps you'd expect from a genuine error. A delivery at `23:50`
and a shipping scan at `00:10` the next day are only 20 minutes apart, but
comparing raw timestamps with a timezone offset can flip which **date**
each one lands on.

Re-running the same comparison at the **date level** removes that artifact:

````python
dq_scan_df = bronze_df \
    .withColumn("_dq_issue",
        when(col("orderid").isNull(),                                       lit("NULL_ORDER_ID"))
        .when(col("customerid").isNull(),                                   lit("NULL_CUSTOMER_ID"))
        .when(col("orderdate").isNull(),                                    lit("NULL_ORDER_DATE"))
        .when(
            col("shippingdate").isNotNull() &
            (col("shippingdate").cast(DateType()) < col("orderdate").cast(DateType())),
            lit("SHIP_BEFORE_ORDER")
        )
        .when(
            col("actualdeliverydate").isNotNull() & col("shippingdate").isNotNull() &
            (col("actualdeliverydate").cast(DateType()) < col("shippingdate").cast(DateType())),
            lit("DELIVERY_BEFORE_SHIP")
        )
        .when(col("orderdate") > current_timestamp(),                       lit("FUTURE_ORDER_DATE"))
        .when(~col("orderchannel").isin(VALID_CHANNELS),                    lit("INVALID_ORDER_CHANNEL"))
        .otherwise(lit(None))
    )

print("=== DQ Issues Found (date-level) ===")
dq_scan_df.groupBy("_dq_issue").count().orderBy("count", ascending=False).display()
````

````python
# Confirm the remaining 114 are a consistent 1-day pattern, not random noise
dq_scan_df.filter(col("_dq_issue") == "DELIVERY_BEFORE_SHIP") \
    .select(
        datediff(col("shippingdate").cast(DateType()), col("actualdeliverydate").cast(DateType())).alias("days_diff")
    ) \
    .groupBy("days_diff").count() \
    .orderBy("days_diff", ascending=False) \
    .display()
````

## Step 5 — Root Cause & Decision: DELIVERY_BEFORE_SHIP

### What we found
At the date level, **114 records** remain — all with exactly a **1-day**
difference. That consistency points to a systemic cause, not random data
entry errors.

### Why it happens
Supabase stores timestamps in UTC; the delivery scan system records in IST
(UTC+5:30). Near midnight, the same real-world moment can land on different
calendar dates once cast down to `DateType()`:

```
Delivered : June 14, 23:45 IST  →  stored as June 14
Shipped   : June 15, 00:30 IST  →  stored as June 15
```
Physically the order shipped before it was delivered — only the date label
flips, not the actual sequence of events.

### Decision → Flag, don't quarantine
We initially quarantined these 114 orders. While building `silver.order_items`,
that decision turned out to orphan **352 downstream line items** — their
parent order no longer existed in `silver.orders`, hiding real
quantity/revenue from Gold reports.

But the defect is confined to two date columns — `customer_id`,
`order_channel`, and the line items themselves were never in question.
Quarantining the *entire order* over a problem in two columns discarded
more than the defect justified, and the cost showed up two tables later.

**We keep all 126,036 orders** and add `_data_note =
'POSSIBLE_TIMEZONE_OFFSET_1DAY'` for the 114 affected rows, so they're
still queryable/excludable for delivery-SLA analysis without breaking
referential integrity for everything downstream.

## Step 6 — Investigate: Is `orderid` Actually Unique?
Bronze is fed by Lakeflow Connect CDC. If an order's status changes, CDC
*could* land a new row instead of updating in place — let's confirm before
assuming a dedup step is needed.

````python
dupe_check_df = bronze_df.groupBy("orderid").count().filter("count > 1")
dupe_count = dupe_check_df.count()
print(f"Distinct orderid values with more than 1 row: {dupe_count:,}")
````

**Finding: 0 duplicates.** Lakeflow Connect is upserting in place for this
table — no dedup step needed.

## Step 7 — Investigate: OrderChannel
Check all distinct channel values before deciding anything.

````python
bronze_df.groupBy("orderchannel").count().orderBy("count", ascending=False).display()
````

**Finding: clean.** `Online` (81,923) and `Retail PoS` (44,113) are both
valid GlobalMart sales channels — no typos, no unknowns. No action needed.

## Step 8 — Context: NULL Shipping & Delivery Dates
The DQ scan does **not** flag these as issues — here's why.

````python
bronze_df.select(
    count(when(col("shippingdate").isNull(), 1)).alias("not_yet_shipped"),
    count(when(col("shippingdate").isNotNull() & col("actualdeliverydate").isNull(), 1)).alias("shipped_not_delivered"),
    count(when(col("actualdeliverydate").isNotNull(), 1)).alias("delivered")
).display()
````

| Column | NULL means | Action |
|---|---|---|
| `shippingdate` | Order placed, not yet shipped | **Keep** — active order |
| `actualdeliverydate` | Shipped, not yet received | **Keep** — in-transit order |
| `expecteddeliverydate` | Estimate not set (older orders) | **Keep** — `is_late` will be null |

Quarantining these would remove every open and in-transit order from Silver.

## Step 9 — Transform

No rows are split out this time — every Bronze row proceeds to Silver. All
126,036 rows get the same business columns; the 114 affected by
`DELIVERY_BEFORE_SHIP` additionally get `_data_note` populated.

| Column | Transformation | Business Need |
|---|---|---|
| `orderdate` / `shippingdate` / `expecteddeliverydate` / `actualdeliverydate` | Cast TIMESTAMP → DateType | Date-only needed for `dim_date` join in Gold |
| `order_to_ship_days` | `datediff(shippingdate, orderdate)` | Fulfillment SLA KPI |
| `ship_to_delivery_days` | `datediff(actualdeliverydate, shippingdate)` | Carrier transit-time KPI |
| `delivery_delay_days` | `datediff(actualdeliverydate, expecteddeliverydate)` | Customer experience KPI — positive = late |
| `is_delivered` | `actualdeliverydate IS NOT NULL` | Boolean flag for Gold queries |
| `is_late` | `actualdeliverydate > expecteddeliverydate` | Drives SLA breach reporting |
| `order_status` | Derived from date columns | Single readable status for Genie/BI |
| `_data_note` | `'POSSIBLE_TIMEZONE_OFFSET_1DAY'` where flagged, else `NULL` | Lets analysts filter out/in the 114 affected rows for delivery-SLA analysis |
| All columns | Rename to lowercase snake_case | Consistent naming convention across all Silver tables |

````python
transformed_df = (
    bronze_df
    .withColumn("orderdate",            col("orderdate").cast(DateType()))
    .withColumn("shippingdate",         col("shippingdate").cast(DateType()))
    .withColumn("expecteddeliverydate", col("expecteddeliverydate").cast(DateType()))
    .withColumn("actualdeliverydate",   col("actualdeliverydate").cast(DateType()))

    .withColumn("order_to_ship_days",
        when(col("shippingdate").isNotNull(),
             datediff(col("shippingdate"), col("orderdate"))
        ).otherwise(lit(None).cast("int"))
    )
    .withColumn("ship_to_delivery_days",
        when(col("actualdeliverydate").isNotNull() & col("shippingdate").isNotNull(),
             datediff(col("actualdeliverydate"), col("shippingdate"))
        ).otherwise(lit(None).cast("int"))
    )
    .withColumn("delivery_delay_days",
        when(col("actualdeliverydate").isNotNull() & col("expecteddeliverydate").isNotNull(),
             datediff(col("actualdeliverydate"), col("expecteddeliverydate"))
        ).otherwise(lit(None).cast("int"))
    )
    .withColumn("is_delivered", col("actualdeliverydate").isNotNull())
    .withColumn("is_late",
        when(
            col("actualdeliverydate").isNotNull() & col("expecteddeliverydate").isNotNull(),
            col("actualdeliverydate") > col("expecteddeliverydate")
        ).otherwise(lit(None).cast("boolean"))
    )
    .withColumn("order_status",
        when(col("actualdeliverydate").isNotNull(), lit("Delivered"))
        .when(col("shippingdate").isNotNull(),      lit("Shipped"))
        .otherwise(                                  lit("Pending"))
    )
    .withColumn("_data_note",
        when(
            col("actualdeliverydate").isNotNull() & col("shippingdate").isNotNull() &
            (col("actualdeliverydate") < col("shippingdate")),
            lit("POSSIBLE_TIMEZONE_OFFSET_1DAY")
        ).otherwise(lit(None).cast("string"))
    )

    .withColumnRenamed("orderid",              "order_id")
    .withColumnRenamed("customerid",           "customer_id")
    .withColumnRenamed("orderdate",            "order_date")
    .withColumnRenamed("shippingdate",         "shipping_date")
    .withColumnRenamed("expecteddeliverydate", "expected_delivery_date")
    .withColumnRenamed("actualdeliverydate",   "actual_delivery_date")
    .withColumnRenamed("shippingtierid",       "shipping_tier_id")
    .withColumnRenamed("supplierid",           "supplier_id")
    .withColumnRenamed("orderchannel",         "order_channel")
)

print(f"Total rows           : {transformed_df.count():,}")
print(f"Rows with _data_note : {transformed_df.filter(col('_data_note').isNotNull()).count():,}")

transformed_df.select(
    "order_id", "customer_id", "order_date", "order_channel", "order_status",
    "shipping_date", "actual_delivery_date", "_data_note"
).display()
````

````python
silver_df = transformed_df \
    .withColumn("_silver_updated_at", current_timestamp()) \
    .select(
        "order_id", "customer_id", "order_date", "order_channel", "order_status",
        "shipping_tier_id", "supplier_id",
        "shipping_date", "expected_delivery_date", "actual_delivery_date",
        "order_to_ship_days", "ship_to_delivery_days", "delivery_delay_days",
        "is_delivered", "is_late", "_data_note", "updated_at", "_silver_updated_at"
    )

print(f"Silver-ready rows : {silver_df.count():,}")
silver_df.display()
````

## Step 10 — SCD1 Write: MERGE into Silver

### Why SCD1 (and not a plain overwrite)?
An order has **one current truth** at any moment — its current status,
current expected delivery date. When Lakeflow CDC tells us an order moved
from `Shipped` to `Delivered`, the right answer is to **overwrite that
row's attributes**, not keep both versions around. That's SCD1.

A `MERGE INTO` keyed on `order_id`:
- **Matched** `order_id` → update all columns to the latest values (overwrite, no history)
- **Unmatched** `order_id` → insert as a new row (new order placed)

This also makes the notebook safe to re-run.

````python
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.silver")

if not spark.catalog.tableExists(SILVER_TABLE):
    silver_df.write.format("delta").saveAsTable(SILVER_TABLE)
    print(f"Created {SILVER_TABLE} — initial load, {silver_df.count():,} rows")
else:
    silver_table = DeltaTable.forName(spark, SILVER_TABLE)
    (silver_table.alias("tgt")
        .merge(silver_df.alias("src"), "tgt.order_id = src.order_id")
        .whenMatchedUpdateAll()
        .whenNotMatchedInsertAll()
        .execute())
    print(f"Merged into {SILVER_TABLE} — SCD1 upsert complete")

print(f"{SILVER_TABLE} row count: {spark.table(SILVER_TABLE).count():,}")
````

## Step 11 — Verify

````python
df = spark.table(SILVER_TABLE)
print(f"silver.orders rows : {df.count():,}")
print(f"Columns            : {df.columns}")

df.groupBy("order_status").count().display()
df.groupBy("order_channel").count().display()
df.filter(col("_data_note").isNotNull()).count()

df.select(
    avg("order_to_ship_days").alias("avg_ship_days"),
    avg("ship_to_delivery_days").alias("avg_delivery_days"),
    count(when(col("is_late") == True, 1)).alias("late_deliveries"),
    count(when(col("is_delivered") == True, 1)).alias("total_delivered"),
    count(when(col("_data_note").isNotNull(), 1)).alias("flagged_timezone_artifact")
).display()
````
