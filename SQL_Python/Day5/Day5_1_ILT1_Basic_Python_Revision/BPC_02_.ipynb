## if-else

```python
score = 30
if score < 40 :
  print("Pass")
else :
  print("Fail")
```

## elif

```python
score = 86
if score >= 90:
  print("Grade A")
elif score >= 80:
  print("Grade B")
elif score >= 70:
  print("Grade C")
elif score >= 60:
  print("Grade D")
elif score >= 50:
  print("Grade E")
else:
  print("Fail")
```

## Nested if-else

```python
score = 68
if score >= 40:
  if score >= 90:
    print("Grade A")
  elif score >= 80:
    print("Grade B")
  elif score >= 70:
    print("Grade C")
  elif score >= 60:
    print("Grade D")
  elif score >= 40:
    print("Grade E")
else:
  print("Fail")
```

## For Loop

```python
for i in range (1, 101):
  print(i)
```

```python
for i in "India":
  print(i)
```

```python
count = 1
for i in "india":
  print(count)
  count += 1
  print(i)
```

## Break & Continue

```python
for i in range(1, 10):
  if i == 5:
    #print(f"Breaking the loop at i = {i}")
    continue
  print(i)
```

```python
for i in range(1,10):
```

```python
for i in range(1,10):
  pass
```

## While Loop

```python
x = 1
while x < 5:
  print(x)
  x += 1
```
