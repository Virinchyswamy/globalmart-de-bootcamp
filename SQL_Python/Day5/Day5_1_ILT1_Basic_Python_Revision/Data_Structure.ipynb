#List

```python
cities = ["Delhi", "Mumbai", "Bengaluru", "Chennai", "Kolkata"]
print(cities)
```

```python
cities[0]
```

```python
cities[0:3]
```

```python
cities[-1]
```

List Methods

```python
len(cities)
```

```python
cities.append("Bhubaneswar")
print(cities)
```

```python
cities.insert(0,345)
print(cities)
```

```python
cities.remove(345)
print(cities)
```

```python
cities.pop()
print(cities)
```

```python
cities2 = ["Gurgaon", "Hyderabad", "Jaipur"]
cities.extend(cities2)
print(cities)
```

```python
for city in cities:
  print(city)
```

```python
for city in cities:
  print(city, end=' ')
```

```python
cities.sort()
print(cities)
```

```python
cities.sort(reverse= True)
print(cities)
```

#Tuple

```python
state = ('Karnataka', 'Odisha', 'Nagaland', 'Punjab')
print(state)
```

```python
state_list = list(state)
state_list.append("Karnataka")
state = tuple(state_list)
print(state)
```

Tuple Method

```python
state.count('Karnataka')
```

```python
state.index('Karnataka')
```

#Dictionary

```python
state_city = {
    "State" : "Rajasthan",
    "City" : "Jaipur"
}
print(state_city)
```

```python
state_city['State']
```

```python
state_city.get('State')
```

```python
state_city.keys()
```

```python
state_city.values()
```

```python
state_city.items()
```

```python
state_city.update({'Zone' : 'West'})
print(state_city)
```

```python
state_city.pop('Zone')
```

```python
for x in state_city:
  print(x)
```

```python
for x in state_city:
  print(state_city[x])
```

Multiple Dictonary

```python
myFamily = {
    "Child1" :{
        "name" : "Emli",
        "year" : 2004
    },
    "Child2" :{
        "name" : "Tobias",
        "year" : 2007
    },
    "Child3" :{
        "name" : "Linus",
        "year" : 2011
    }
}
print(myFamily)
```

```python
print(myFamily["Child2"]["name"])
```

```python
for x in myFamily.items():
  print(x)
```
