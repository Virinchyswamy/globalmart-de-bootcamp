## **TASK 1**

You are managing the online sales of three popular products at GlobalMart

Each product has a fixed price and daily sales (units sold) that need to be tracked.

The task is to compute the total income from the sales of each product and the overall income for the day.

```python
# Product prices
price_smartphone = 299.99
price_laptop = 899.99
price_tablet = 199.99

# Units sold today
units_sold_smartphone = 9
units_sold_laptop = 5
units_sold_tablet = 7

# Calculate sales for each product
sales_smartphone = units_sold_smartphone * price_smartphone
sales_laptop = units_sold_laptop * price_laptop
sales_tablet = units_sold_tablet * price_tablet

# Calculate total sales for the day
total_sales = sales_smartphone + sales_laptop + sales_tablet

# Display the sales data
print(f"Today's income from smartphones: {sales_smartphone}")
print(f"Today's income from laptops: {sales_laptop}")
print(f"Today's income from tablets: {sales_tablet}")
print(f"Total sales for the day: {total_sales : .3f}")
```

## **TASK 2:**

GlobalMart offers a discount on a product based on the amount spent by the customer.

Write a Python program that asks the user for the total price of their shopping cart.

The program should then apply a discount based on the following rules:

- If the total price is less than $50, no discount is applied.

- If the total price is between 50 AND 100, a 10% discount is applied.

- If the total price is more than $100, a 20% discount is applied.

The program should display the original price, the discount amount, and the final price after the discount.

```python
# Get the total price from the user (Should be a float value)
total_price = float(input("Enter the total price of your shopping cart: "))

# Initialize the discount rate
discount_rate = 0

# Determine the discount rate based on the total price
if 50 <= total_price <= 100:
    discount_rate = 10
elif total_price > 100:
    discount_rate = 20

# Calculate the discount amount
discount_amount = (discount_rate / 100) * total_price

# Calculate the final price after the discount
final_price = total_price - discount_amount

# Print the original price, discount amount, and final price (Display upto 2 decimal places)
print(f"Original Price: ${total_price:.2f}")
print(f"Discount Applied: ${discount_amount:.2f} ({discount_rate}%)")
print(f"Final Price after Discount: ${final_price:.2f}")
```

## **Task 3:**

Task 3:

GlobalMart is analyzing the profitability of the iPhone 15.
Write a Python program that calculates the profit made on each iPhone 15 sold and the profit margin as a percentage. The cost price of the iPhone 15 is $675.15, and the selling price is 799.99. Use the following formula to calculate the profit margin:



Profit Margin=(Profit/Selling Price)×100

```python
# Set the cost price and selling price for iPhone 15
cost_price = 675.15
selling_price = 799.99

# Calculate the profit
profit = selling_price - cost_price

# Calculate the profit margin as a percentage
profit_margin = (profit / selling_price) * 100

# Print the results
print("Product: iPhone 15")
print(f"Cost Price: ${cost_price:.2f}")
print(f"Selling Price: ${selling_price:.2f}")
print(f"Profit: ${profit:.2f}")
print(f"Profit Margin: {profit_margin:.2f}%")
```

## **TASK 4:**

GlobalMart has different return policies for furniture based on the type of item.

Write a Python program that asks the user for the type of furniture (choose from "sofa", "table", or "chair") and the number of days since the item was purchased.

The return policies are:

- "sofa" can be returned within 45 days.
- "table" can be returned within 30 days.
- "chair" can be returned within 15 days.

The program should check if the furniture item is returnable based on these conditions and inform the user of the return status.

```python
# Get input from the user
furniture_type = input("Enter the type of furniture (sofa, table, chair): ")
days_since_purchase = int(input("Enter the number of days since the item was purchased: "))

# Determine the return eligibility
if furniture_type == "sofa" and days_since_purchase <= 45:
    print("Your return is accepted.")
elif furniture_type == "table" and days_since_purchase <= 30:
    print("Your return is accepted.")
elif furniture_type == "chair" and days_since_purchase <= 15:
    print("Your return is accepted.")
else:
    print("Your return period has expired.")
```

## **Task 5:**

GlobalMart assists customers in creating secure passwords through a password analysis tool.

Write a Python program that prompts customers to propose a password.

The program analyzes the password by counting the number of letters, digits, and special characters (defined as any character that isn't a letter or digit).

To be considered strong, a password must meet the following criteria:

At least 8 characters in total.
A minimum of 3 letters.
At least 2 digits.
At least 1 special character.

```python
# Take string input from the user for a password suggestion
user_input = input("Please enter your password suggestion: ")

# Declare variables to store the count of letters, digits, and special characters
digits = 0
letters = 0
special_chars = 0

# Iterate through each character in the input string
for char in user_input:
    # Check if the character is a digit
    if char.isdigit():
        digits += 1
    # Check if the character is a letter
    elif char.isalpha():
        letters += 1
    # Assume any other character is a special character
    else:
        special_chars += 1

# Print the analysis results
print(f"The input password '{user_input}' has {letters} letters, {digits} digits, and {special_chars} special characters.")

# Check if the password meets the strength criteria
if len(user_input) >= 8 and digits >= 2 and letters >= 3 and special_chars >= 1:
    print("This is a strong password.")
else:
    print("Your password could be stronger. Ensure it has at least 8 characters, with a minimum of 3 letters, 2 digits, and 1 special character.")
```

## **Task 6:**

GlobalMart wants to enhance the checkout experience by allowing customers to enter discount codes for their purchases.

However, these codes must be verified to ensure they are valid before the discount can be applied.

Write a program that will simulate the entry and verification of a discount code, which must match a pre-defined valid code.

The user is prompted to enter a discount code, and if it does not match the valid code, they are prompted to try again.

```python
# Define the valid discount code
valid_discount_code = "SAVE20"

# Prompt the user and verify the discount code
while True:
    # Ask the user to enter a discount code
    entered_code = input("Please enter your discount code: ")

    # Check if the entered code is valid
    if entered_code == valid_discount_code:
        print("Congratulations! Your discount code is valid. Your discount will be applied.")
        break  # Exit the loop as the valid code is entered
    else:
        print("Invalid discount code. Please try again.")
```
